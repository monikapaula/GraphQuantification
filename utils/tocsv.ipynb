```python
import json
import pandas as pd
from pathlib import Path
from utils.data_loader import DATA_ROOT

file_path = file_path = Path(DATA_ROOT) / "split_data" / "deezer_europe" / "split_5" / "train.json"
output_dir = Path("gephicsv")
output_dir.mkdir(exist_ok=True)

with open(file_path, 'r') as f:
    data = json.load(f)

df = pd.DataFrame({
    'Id': data,
    'is_training': 1
})
output_file = output_dir / "deezer_gephi_split_5.csv"
df.to_csv(output_file, index=False)

print(f"Successfully processed {len(df)} training nodes. File saved as 'gephi_training_flag.csv'.")
```
